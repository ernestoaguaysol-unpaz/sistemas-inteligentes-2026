```python
import numpy as np

import tensorflow as tf

import matplotlib.pyplot as plt
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
y = np.array([
    0,
    1,
    1,
    0
])
```

```python
# Creación de un modelo usando la API Sequential
# Una capa Densa seguida de otra
modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(2,), activation="sigmoid")
])
```

```python
# Compilar el modelo: definir un optimizador, función de pérdida y métricas.
modelo.compile(
    # https://www.tensorflow.org/api_docs/python/tf/keras/optimizers/SGD
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
    loss="binary_crossentropy",
    metrics=["accuracy"],
)
```

```python
modelo.summary()
```

```python
modelo.layers
```

```python
modelo.layers[0]
```

```python
modelo.layers[0].get_weights()
```

```python
# Ver con distintos batch_size cómo avanza cada epoch
modelo.fit(X, y, batch_size=1, epochs=10)
```

```python
modelo_fit = modelo.fit(X, y, batch_size=1, epochs=5000, verbose=0)
```

```python
plt.plot(modelo_fit.history['loss'])
```

```python
plt.plot(modelo_fit.history['accuracy'])
```

```python
modelo.layers[0].get_weights()
```

```python
w1 = modelo.layers[0].get_weights()[0][0][0]
w2 = modelo.layers[0].get_weights()[0][1][0]
w0 = modelo.layers[0].get_weights()[1][0]
w0, w1, w2
```

```python
a = -w1 / w2
b = -w0 / w2
a, b
```

```python
plt.axline((0, b), slope=a, color='C0', label='by slope')
plt.scatter(x=X[:, 0],y=X[:, 1],c=y)
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.grid()
plt.show()
```

```python
modelo.predict(X)
```

```python
modelo.evaluate(X, y)
```
