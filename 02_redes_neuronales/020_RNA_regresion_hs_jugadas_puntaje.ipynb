```python
# Se importan las librerías generales necesarias
import pandas as pd
from sklearn.model_selection import train_test_split
import tensorflow as tf
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
dataset = pd.read_csv('dataset_hs_jugadas_puntaje.csv')
dataset.head()
```

```python
sc = StandardScaler()
sc.fit(dataset[['HsJuego', 'Puntaje']])
dataset[['HsJuego', 'Puntaje']] = sc.transform(dataset[['HsJuego', 'Puntaje']])
dataset.head()
```

```python
X = dataset.iloc[:, 0:1]
y = dataset.iloc[:, 1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
# Construir el modelo de red neuronal
modelo = tf.keras.Sequential([
    tf.keras.layers.Input((1,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1)
])
```

```python
modelo.compile(
    # Elección del optmizador: algoritmo GD con la cual red va a aprender.
    optimizer=tf.keras.optimizers.SGD(
        learning_rate=0.001
    ),
    # Pérdida del modelo
    loss=tf.keras.losses.MSE,
    # Métricas para luego evaluar la evolución del aprendizaje
    metrics=[tf.keras.metrics.MSE],
)
```

```python
# Entrenar el modelo
historial = modelo.fit(
    X_train,
    y_train,
    epochs=500,
    verbose=0
)
```

```python
print(historial.history.keys())
```

```python
plt.title('Historial de rendimiento MSE')
plt.plot(historial.history['mean_squared_error'], label="MSE")
plt.xlabel('Epochs')
plt.ylabel('Rendimiento MSE')
plt.legend()
plt.show()
```

```python
plt.title('Historial de pérdida MSE')
plt.plot(historial.history['loss'], label="Pérdida MSE")
plt.xlabel('Epochs')
plt.ylabel('Pérdida MSE')
plt.legend()
plt.show()
```

```python
# Evaluar el modelo con conjunto de entrenamiento
modelo.evaluate(X_train, y_train)
```

```python
# Evaluar el modelo con conjunto de prueba
modelo.evaluate(X_test, y_test)
```

```python
# Hacer predicciones
y_pred = modelo.predict(X_test)
```
